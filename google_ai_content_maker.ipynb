## 🔑 Налаштування / Konfiguracja

| 🇺🇦 | 🇵🇱 |
| --- | --- |
| **Colab:** Відкрийте `🔑 Secrets` (ліва панель) → додайте секрет `GOOGLE_API_KEY` → вставте ваш ключ з [Google AI Studio](https://aistudio.google.com/app/apikey). | **Colab:** Otwórz `🔑 Secrets` (lewy panel) → dodaj sekret `GOOGLE_API_KEY` → wklej klucz z [Google AI Studio](https://aistudio.google.com/app/apikey). |
| **Локально:** Встановіть змінну оточення `export GOOGLE_API_KEY=ваш_ключ` | **Lokalnie:** Ustaw zmienną środowiskową `export GOOGLE_API_KEY=twój_klucz` |

```python
import os
import google.generativeai as genai

# Load API key: Colab Secrets first, then environment variable
try:
    from google.colab import userdata
    GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
except Exception:
    GOOGLE_API_KEY = os.environ.get('GOOGLE_API_KEY', '')

if not GOOGLE_API_KEY:
    raise ValueError("GOOGLE_API_KEY not set. See setup instructions above.")

genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel("gemini-1.5-flash")
print("✅ Gemini configured / Gemini skonfigurowany")
```

## ✍️ Генерація тексту · Generowanie tekstu
### Gemini 1.5 Flash

| 🇺🇦 | 🇵🇱 |
| --- | --- |
| Функція `generate_content` приймає тему, формат і мову, після чого викликає Gemini API та повертає готовий контент. | Funkcja `generate_content` przyjmuje temat, format i język, a następnie wywołuje Gemini API i zwraca gotową treść. |

```python
FORMATS = {
    "linkedin_post": "a professional LinkedIn post (max 300 words, include 3 relevant hashtags)",
    "blog_intro": "a compelling blog post introduction (2-3 paragraphs)",
    "cold_email": "a short, personalized cold outreach email (subject line + body, max 150 words)",
    "veo_prompt": "a detailed visual scene description for a 15-second AI-generated video (Veo style: camera angle, lighting, motion, mood)",
}

LANG_LABELS = {"uk": "Ukrainian", "pl": "Polish"}

def generate_content(topic: str, fmt: str, lang: str = "uk") -> str:
    """Generate content for a given topic, format and language."""
    lang_label = LANG_LABELS.get(lang, "Ukrainian")
    fmt_desc = FORMATS.get(fmt, fmt)
    prompt = (
        f"You are a professional content creator. "
        f"Write {fmt_desc} about the following topic: '{topic}'. "
        f"Write entirely in {lang_label}. Be concise, engaging, and professional."
    )
    response = model.generate_content(prompt)
    return response.text
```

### 📣 LinkedIn Post — 🇺🇦 Українська

```python
topic = "No-code automation for small business with Make.com"

post_ua = generate_content(topic, fmt="linkedin_post", lang="uk")
print("🇺🇦 LinkedIn Post:\n")
print(post_ua)
```

### 📣 LinkedIn Post — 🇵🇱 Polski

```python
post_pl = generate_content(topic, fmt="linkedin_post", lang="pl")
print("🇵🇱 LinkedIn Post:\n")
print(post_pl)
```

## 🎬 Veo — AI Video Generation

| 🇺🇦 | 🇵🇱 |
| --- | --- |
| **Veo** від Google DeepMind генерує відео до 60 секунд з детального текстового опису. Прямий API поки доступний через [Google Cloud Vertex AI](https://cloud.google.com/vertex-ai/generative-ai/docs/video/generate-videos) у обмеженому Preview. | **Veo** od Google DeepMind generuje wideo do 60 sekund ze szczegółowego opisu tekstowego. Bezpośrednie API dostępne przez [Google Cloud Vertex AI](https://cloud.google.com/vertex-ai/generative-ai/docs/video/generate-videos) w ograniczonym Preview. |
| **Workflow:** Gemini генерує детальний prompt для Veo → Veo рендерить відео → готово для соцмереж. | **Workflow:** Gemini generuje szczegółowy prompt dla Veo → Veo renderuje wideo → gotowe do social media. |

```python
# Use Gemini to craft a Veo-ready video prompt
video_topic = "A no-code automation workflow saving time for a modern entrepreneur"

veo_prompt_ua = generate_content(video_topic, fmt="veo_prompt", lang="uk")
veo_prompt_pl = generate_content(video_topic, fmt="veo_prompt", lang="pl")

print("🇺🇦 Veo Prompt (for Google Veo video generation):\n")
print(veo_prompt_ua)
print("\n" + "─" * 60 + "\n")
print("🇵🇱 Veo Prompt:\n")
print(veo_prompt_pl)

print("\n💡 Copy these prompts into Veo (Google Cloud Vertex AI) to generate your video.")
```

## 🚀 Batch Content Generation

| 🇺🇦 | 🇵🇱 |
| --- | --- |
| Генерація контенту для списку тем одразу двома мовами — основа для масштабованого контент-маркетингу. | Generowanie treści dla listy tematów od razu w dwóch językach — podstawa skalowalnego content marketingu. |

```python
topics = [
    "Make.com webhooks for CRM automation",
    "AI agents for HR onboarding",
    "n8n vs Zapier: which to choose in 2025",
]

results = {}
for t in topics:
    results[t] = {
        "ua": generate_content(t, fmt="linkedin_post", lang="uk"),
        "pl": generate_content(t, fmt="linkedin_post", lang="pl"),
    }
    print(f"✅ Generated: {t}")

print("\n🎉 Batch complete / Batch zakończony!")
```

## 💾 Export to CSV

```python
import csv
import io

output = io.StringIO()
writer = csv.writer(output, quoting=csv.QUOTE_ALL)
writer.writerow(["topic", "lang", "format", "content"])

for t, langs in results.items():
    for lang_code, content in langs.items():
        writer.writerow([t, lang_code, "linkedin_post", content])

csv_content = output.getvalue()

# Save to file
with open("content_output.csv", "w", encoding="utf-8") as f:
    f.write(csv_content)

print("✅ Saved to content_output.csv")
print(f"📊 Total rows: {len(results) * 2}")
print("\nPreview (first 500 chars):")
print(csv_content[:500])
```
